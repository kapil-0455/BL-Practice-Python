## Functions (`def`)
- it is a block of code which can run whenever we called.
- You can pass data, known as `parameters`, into a function.
-  Make our code modular.

```python
def greet(name : str):
    '''Greet Someone your friend'''
    print(f"Hello {name}! How are you ?")


name = input('Enter a name : ')
greet(name)
```

```python
# Use of validation 
def greet(name : str):
    '''Greet Someone but give only string'''
    if name.isalpha():
        print(f"Hello {name}! How are you ?")
    else:
        print('Please enter a valid characters')

name = input('Enter a name : ')
greet(name)
```
