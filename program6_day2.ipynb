# String Formatting Expression

- Used to create **clean, readable, and well-aligned** output.
- Written inside an **f-string**.

### Common Formatting Expressions

- `^` -> Center align text.
- `<` -> Left align text.
- `>` -> Right align text.
- `* n` -> Repeat a character `n` times.
- `" " * n` -> Repeat spaces `n` times.
- `.2f` -> Display a float with **2 decimal places**.
- `+` -> Concatenate (join) strings.


```python
#Example

price = 20

# Center align text within 10 characters
print(f"{'Item Code':^10}")

# Repeat '-' 74 times
print("-" * 74)

# Repeat space 54 times
print(" " * 54 + "Total Amount")

# Display float with 2 decimal places
print(f"{price:.2f}")

# String concatenation
print("₹" + str(price))
```

```python
# Practice code Bill generation


#Setting Constants for Fruits
MANGO_GST = 0.12      # 12% GST
BANANA_GST = 0.05     # 5% GST

#Enter Buyer Name, Price per kg, and Quantity
buyer_name = input("Enter Buyer Name: ")

mango_price_kg = int(input("Enter Mango price per kg: "))
mango_quantity_kg = float(input("Enter Mango Quantity in kg: "))

banana_price_kg = int(input("Enter Banana price per kg: "))
banana_quantity_kg = float(input("Enter Banana Quantity in kg: "))

# Calculate Total Price
total_price_mango = mango_price_kg * mango_quantity_kg
total_price_banana = banana_price_kg * banana_quantity_kg

# Calculate GST
total_gst_mango = total_price_mango * MANGO_GST
total_gst_banana = total_price_banana * BANANA_GST

# Calculate Billing Amount (Including GST)
total_billing_mango = total_price_mango + total_gst_mango
total_billing_banana = total_price_banana + total_gst_banana

# Calculate Total Amount and Round It
total_amount = total_billing_mango + total_billing_banana
total_round_amount = round(total_amount)

# Print the Bill
print(f"\nBuyer Name: {buyer_name}")
print(f"{'-' * 74}")

print(
    f"| {'Item Code':^10} | {'Price/Unit':^10} | {'# unit':^5}",
    f"| {'Price':^5} | {'GST':^10} | {'Total w/ GST':^10} |"
)

print(f"{'-' * 74}")

print(
    f"| {'Mango':^10} | {'Rs ' + str(mango_price_kg):^10} | {mango_quantity_kg:^5}",
    f"| {'Rs ' + str(total_price_mango):^5} | {'Rs ' + str(total_gst_mango):^10}",
    f"| {'Rs ' + str(total_billing_mango):^10} |"
)

print(
    f"| {'Banana':^10} | {'Rs ' + str(banana_price_kg):^10} | {banana_quantity_kg:^5}",
    f"| {'Rs ' + str(total_price_banana):^5} | {'Rs ' + str(total_gst_banana):^10}",
    f"| {'Rs ' + str(total_billing_banana):^10} |"
)

print(f"{'-' * 74}")

print(f"Total {' ' * 54} \u20B9 {total_amount:.2f}")
print(f"Total Round {' ' * 48} \u20B9 {total_round_amount:.2f}")

print(f"{'-' * 74}")
```
