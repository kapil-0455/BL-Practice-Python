## Convert String Into DateObject

## `strptime()` – String to Datetime

- **Full Form:** **Str**ing **P**arse **Time**
- Converts a **string** into a **`datetime` object**.
- Requires the input string and its matching format codes.

```python
import datetime

date_str = "22 Jun 2024"
# specifying the data 
# %d -> first thing is date and likewise
date_obj = datetime.datetime.strptime(date_str, "%d %b %Y")

print(date_obj)
print(type(date_obj))
```

## `strftime()` – Datetime to String

- **Full Form:** **Str**ing **F**ormat **Time**
- Converts a **`datetime` object** into a **formatted string**.
- Used when displaying or storing dates in a desired format.


```python
import datetime

current_date = datetime.datetime.now()

formatted_date = current_date.strftime("%d %b %Y")

print(formatted_date)
print(type(formatted_date))
```
