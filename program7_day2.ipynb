```python
# Practice Program 7: School Report
SCHOOL_NAME = "New School Of Learning"
CLASS_NAME = "Class XI"
TOTAL_SUBJECT_MARKS = 100
TOTAL_MARKS = TOTAL_SUBJECT_MARKS * 3
NUMBER_STUDENTS = 1

def generate_report(student_no):
    #Details
    student_name = input(f"\nEnter name of Student {student_no}: ")
    phys_marks = int(input(f"Enter Physics marks out of 100 for Student {student_no}: "))
    chem_marks = int(input(f"Enter Chemistry marks out of 100 for Student {student_no}: "))
    math_marks = int(input(f"Enter Mathematics marks out of 100 for Student {student_no}: "))

    total_student_marks = phys_marks + chem_marks + math_marks

    phys_perc = round((phys_marks / TOTAL_SUBJECT_MARKS) * 100, 2)
    chem_perc = round((chem_marks / TOTAL_SUBJECT_MARKS) * 100, 2)
    math_perc = round((math_marks / TOTAL_SUBJECT_MARKS) * 100, 2)
    overall_perc = round((total_student_marks / TOTAL_MARKS) * 100, 2)

    print(f"\n{SCHOOL_NAME} - {CLASS_NAME} - {student_name}")

    print("-" * 72)
    print(f"| {'Subject':^15} | {'Total Marks':^15} | {'Marks Obtained':^15} | {'Percentage':^15} |")
    print("-" * 72)
    print(f"| {'Physics':^15} | {TOTAL_SUBJECT_MARKS:^15} | {phys_marks:^15} | {phys_perc:^15} |")
    print(f"| {'Chemistry':^15} | {TOTAL_SUBJECT_MARKS:^15} | {chem_marks:^15} | {chem_perc:^15} |")
    print(f"| {'Mathematics':^15} | {TOTAL_SUBJECT_MARKS:^15} | {math_marks:^15} | {math_perc:^15} |")
    print("-" * 72)

    print(f"| {'Total':^15} | {TOTAL_MARKS:^15} | {total_student_marks:^15} | {overall_perc:^15} |")
    print("-" * 72)
    print()

for student in range(1, NUMBER_STUDENTS + 1):
    generate_report(student)
```
