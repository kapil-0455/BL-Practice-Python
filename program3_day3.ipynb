###### `len()` -> find the length of list
###### `sum()` -> find the sum of all elements 

```python
# Prcatice problem calculate average 

def calculate_average(num_lis):
    if len(num_lis) == 0 :
        return 0
    
    return sum(num_lis) / len(num_lis)

num_lis = [10]
print(f"Original List : {num_lis}")

average = calculate_average(num_lis)

print(f"Average of list {average}")
```
