# Python Data Types

- **Data Type:** Defines the type of data stored in a variable.
- **Implicit Assignment:** Python automatically assigns the data type based on the assigned value.
- **Classes & Objects:** Every data type is a **class**, and every variable is an **object**.

### **Basic Data Types**
- `int`
- `float`
- `bool`
- `str`

### **Derived Data Types (Collections)**
- `list`
- `tuple`
- `set`
- `dict`

### **Typecasting (Explicit Conversion)**
- Convert one data type to another using:
  - `int()`
  - `float()`
  - `str()`

---

## **Python Constants**

- Values that **should not change** throughout the code.
- Usually declared at the **top** of the file.
- Can be **global**.
- Written in **UPPER_CASE**.

Example:

```python
PI = 3.14159
MAX_USERS = 100
```

---

## **Python Multiplication & Power Operators**

- **Multiplication** -> `*`
- **Power (Exponent)** -> `**`

Example:

```python
5 * 4      # 20
2 ** 3     # 8
```

---

## **Python Operator Precedence (PEMDAS)**

Python evaluates expressions in the following order:

- **P** -> Parentheses `()`
- **E** -> Exponents `**`
- **M** -> Multiplication `*`
- **D** -> Division `/`
- **A** -> Addition `+`
- **S** -> Subtraction `-`

> **Note:** Operators with the same precedence are evaluated **from left to right**.

Example:

```python
2 + 3 * 4 ** 2
# Output: 50
```

````python
# python constants 
PI = 22 / 7

radius_inch = int(input('Enter your radius in inch : '))
radius_cm = radius_inch * 2.48

circumference = 2 * PI * radius_cm 
area = PI * (radius_cm ** 2)

print(f"Area of a circle is {area}.", f"Circumference of a circle is {circumference}" , sep='\n' )
````

# Float Format Specifier (`f-string`)

- **Purpose:** Formats a floating-point number for clean and readable output.
- **Syntax:** `f"{variable:.2f}"`
- **`:` (Colon):** Separates the variable from the format specifier.
- **`.2`:** Displays **2 decimal places**.
- **`f`:** Specifies floating-point formatting.
- **Rounding:** Python automatically rounds to the specified decimal places.

**Example:**
```python
pi = 3.14159
print(f"{pi:.2f}")   # Output: 3.14
```

````python
# python constants 
PI = 22 / 7

radius_inch = int(input('Enter your radius in inch : '))
radius_cm = radius_inch * 2.48

circumference = 2 * PI * radius_cm 
area = PI * (radius_cm ** 2)

print(f"Area of a circle is {area:.2f}.", f"Circumference of a circle is {circumference:.2f}" , sep='\n' )
````
