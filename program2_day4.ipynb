# datetime module handles dates and times.

- Distinct Modules: Date and time aren't built-in types in Python. Separate modules — datetime and time  provide this functionality.

- **Current Date & Time**: Import datetime, then call `datetime.now()`.
- Time Zones: `datetime.now()` without a location returns **`global/system time`**.
- **Formatting with strftime()**: Formats and extracts date/time components using `% directives`.

---

## Main Classes in `datetime`

- **`date`**  
  Stores only the **date** (year, month, day).  
  **Example:** `2026-07-25`

- **`time`**  
  Stores only the **time** (hour, minute, second, microsecond).  
  **Example:** `14:30:45`

- **`datetime`**  
  Stores both **date and time** together.  
  **Example:** `2026-07-25 14:30:45`

- **`timedelta`**  
  Represents the **difference between two dates or times**. It is also used to add or subtract days, hours, minutes, etc.  
  **Example:** `timedelta(days=5)`

- **`timezone`**  
  Represents a **time zone** and is used to create timezone-aware date and time objects.  
  **Example:** `UTC`, `UTC+05:30`

#### **`.strftime()`** Formated codes

- **`%B`** -> Full month name  
  **Example:** `June`

- **`%b`** -> Abbreviated month name  
  **Example:** `Jun`

- **`%A`** -> Full weekday name  
  **Example:** `Tuesday`

- **`%a`** -> Abbreviated weekday name  
  **Example:** `Tue`

- **`%Y`** -> Four-digit year  
  **Example:** `2024`

- **`%y`** -> Two-digit year  
  **Example:** `24`

- **`%d`** -> Day of the month (01–31)  
  **Example:** `15`

- **`%H`** -> Hour (24-hour format)  
  **Example:** `18`

- **`%I`** -> Hour (12-hour format)  
  **Example:** `06`

- **`%M`** -> Minutes (00–59)  
  **Example:** `45`

- **`%S`** -> Seconds (00–59)  
  **Example:** `30`

- **`%p`** -> AM or PM  
  **Example:** `AM` or `PM`




```python
import datetime

curr_date = datetime.datetime.now()
print(f'Current date and time is {curr_date}')

print(f'Full Month name : {curr_date.strftime('%B')}')
print(f"Abbreviated Month name : {curr_date.strftime('%b')}")
print(f"Month in number (01 - 12) : {curr_date.strftime('%m')}")
print(f"Weekday of that month : {curr_date.strftime('%A')}")
print(f"Abbreviated Weekday of that month : {curr_date.strftime('%a')}")
print(f"Year (four digit) : {curr_date.strftime('%Y')}")
print(f"Year (two digit) : {curr_date.strftime('%y')}")
print(f'Day of the month (01 - 31) : {curr_date.strftime('%d')}')
print(f'Print hour of the day (24 hour clock) : {curr_date.strftime('%H')}')
print(f'Print hour of the day (12 hour clock) : {curr_date.strftime('%I')}')
print(f'Minutes (00 - 59) : {curr_date.strftime('%M')}')
print(f"Second (0-60) : {curr_date.strftime('%S')}")
print(f"AM/PM indicator : {curr_date.strftime('%p')}")
```

---

#### For specific region (such as Asia/Kolkata)
- we can use a separate Python library called **`pytz (Python Time Zone)`** to declare a time zone and pass it into the function

```python
import datetime
import pytz

ist = pytz.timezone('Asia/Kolkata')


curr_date = datetime.datetime.now(ist)  
print(f'Current date and time is {curr_date}')

print(f'Full Month name : {curr_date.strftime('%B')}')
print(f"Abbreviated Month name : {curr_date.strftime('%b')}")
print(f"Month in number (01 - 12) : {curr_date.strftime('%m')}")
print(f"Weekday of that month : {curr_date.strftime('%A')}")
print(f"Abbreviated Weekday of that month : {curr_date.strftime('%a')}")
print(f"Year (four digit) : {curr_date.strftime('%Y')}")
print(f"Year (two digit) : {curr_date.strftime('%y')}")
print(f'Day of the month (01 - 31) : {curr_date.strftime('%d')}')
print(f'Print hour of the day (24 hour clock) : {curr_date.strftime('%H')}')
print(f'Print hour of the day (12 hour clock) : {curr_date.strftime('%I')}')
print(f'Minutes (00 - 59) : {curr_date.strftime('%M')}')
print(f"Second (0-60) : {curr_date.strftime('%S')}")
print(f"AM/PM indicator : {curr_date.strftime('%p')}")
```
