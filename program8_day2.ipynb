```python
# PROG 8: Retail Invoicing App

HG_GST = 5
PF_GST = 12

HG_ITEM = "Household Good"
PF_ITEM = "Processed Food"

def generate_invoice():

    buyer = input("Enter Buyer Name: ")

    hg_price = float(input("Enter Household Good Price: "))
    hg_qty = float(input("Enter Household Good Quantity: "))

    pf_price = float(input("Enter Processed Food Price: "))
    pf_qty = float(input("Enter Processed Food Quantity: "))

    hg_total = hg_price * hg_qty
    hg_gst = hg_total * HG_GST / 100
    hg_final = hg_total + hg_gst

    pf_total = pf_price * pf_qty
    pf_gst = pf_total * PF_GST / 100
    pf_final = pf_total + pf_gst

    total = hg_final + pf_final
    total_round = round(total)

    print(f"\nBuyer Name: {buyer}")

    print("-" * 82)
    print(f"| {'Item Code':^18} | {'Price/Unit':^10} | {'#Unit':^6} | {'Price':^10} | {'GST':^10} | {'Total w/ GST':^14} |")
    print("-" * 82)
    print(f"| {HG_ITEM:^18} | {'Rs ' + str(hg_price):^10} | {hg_qty:^6} | {'Rs ' + str(hg_total):^10} | {'Rs ' + str(hg_gst):^10} | {'Rs ' + str(hg_final):^14} |")
    print(f"| {PF_ITEM:^18} | {'Rs ' + str(pf_price):^10} | {pf_qty:^6} | {'Rs ' + str(pf_total):^10} | {'Rs ' + str(pf_gst):^10} | {'Rs ' + str(pf_final):^14} |")
    print("-" * 82)
    print(f"{'Total':<70} ₹ {total:.2f}")
    print(f"{'Total Round':<70} ₹ {total_round:.2f}")
    print("-" * 82)

generate_invoice()
```
