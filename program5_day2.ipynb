# Python Floor Division (`//`)

- Returns the **quotient** (whole number).
- Rounds the result **down** to the nearest integer.



```python
#Example 
print(5//3) # round of the value
```

# Python Modulo Operator (`%`)

- Returns the **remainder** of a division.

```python
#Example 
print(5 % 2)
```

# Integer Format Specifier (`d`)

- Used to format **integers** in an f-string.
- **Syntax:** `f"{number:02d}"`

Where:
- `:` → Separates the variable from the format specifier.
- `02` → Displays at least **2 digits** (adds leading `0` if needed).
- `d` → Formats the value as an **integer**.

```python
# Example
number = 7
print(f"{number:02d}")  # :0=2d  internally treated like this
print(f"{number:03d}")  

# Here, 22 means minimum width = 22. Since there is no leading 0, Python pads with spaces.
print(f"{number:22d}")

# {value:[fill][align][width]d} -> this is the formula

```

```python
# testcases of Integer format specifier
number = 7

print(f"{number:0>5d}")   # 00007
print(f"{number:*>5d}")   # ****7
print(f"{number:#>5d}")   # ####7
print(f"{number:_>5d}")   # ____7
print(f"{number:.>5d}")   # ....7
print(f"{number:-<5d}")   # 7----
print(f"{number:*^5d}")   # **7**
```

```python
# Practice example -> Time conversion
total_seconds = int(input('Enter time in seconds : '))

hours = total_seconds // 3600
minutes = total_seconds % 3600 // 60
seconds = total_seconds % 60


print(f'Total time Duration : {total_seconds}' , f"seconds in HH:MM:SS format : {hours:02d}:{minutes:02d}:{seconds:02d}" , sep='\n')
```
