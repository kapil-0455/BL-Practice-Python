### Built-in function `input()`
- recieve user input from standard input. when return pressed user input is captured.

#### Python Help Function
- `help()` function is used to look up detailed documentation.

#### Formatted string literal, or **f-string**
- evaluate the expression.
- convert it into string.
- change int that particular level.
- thats why performance is much better than printing using variables. 

#### print(value, ..., `sep=' ', end='\n'`, file=sys.stdout, flush=False)
- using help function we can see how many function parameters print have.
- by default sep is set to ' ' and end is set to next line.
- we can set by own customize it .




```python
# Using help() to know more about print and input
help(print)
help(input)

# Step 1: Collect Employee Details
employee_name = input("Enter Employee Name: ")
employee_id = input("Enter Employee ID: ")
department = input("Enter Department: ")
print("\nEmployee Details", employee_name, employee_id,department,sep=" | ",end="\n")
print(f"Welcome {employee_name}!")
print(f"Your Employee ID is {employee_id}.")
print(f"You have joined the {department} department.")
```
