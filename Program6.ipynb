# Exception Handling


```python
#Error -> An error is a problem that causes your program to stop or behave unexpectedly.
# How to Handle that ? -> using exception handling 
# we use throw and except 
# we write those code inside it that may give error 

# PROG 6.1 Handling User Error


def calculate_sum(x: int, y: int):
    """
    Returns the sum of two numbers.
    """
    return x + y


# Handling bad user inputs
# Bad: Not handling the exact error occurred

try:
    total_sum = int(input('Enter value for Total Sum '))
    bonus_points = int(input('Enter value for Bonus Points '))

    final_score = calculate_sum(total_sum, bonus_points)

    print(f"Final Score is {final_score}")

except:
    print("Try Again and Enter a Proper Value")
```

Exact user error 

```python

def calculate_sum(x: int, y: int):
    """
    Returns the sum of two numbers.
    """
    return x + y


# Handling the exact error occurred

try:
    total_sum = int(input('Enter value for Total Sum '))
    bonus_points = int(input('Enter value for Bonus Points '))


    final_score = calculate_sum(total_sum, bonus_points)

    print(f"Final Score is {final_score}")

except ValueError as err:
    print(f"Error Occured you give wrong type: {err}")

except:
    print("Try Again and Enter a Proper Value")
```

# Use of Finally Block

```python
# basic real world example of finally is 
# use case 
# used in file Handling open the final do work and at last close the file
def calculate_sum(x: int, y: int):
    """
    Returns the sum of two numbers.
    """
    return x + y


# Handling the exact error occurred

try:
    total_sum = int(input('Enter value for Total Sum '))
    bonus_points = int(input('Enter value for Bonus Points '))


    final_score = calculate_sum(total_sum, bonus_points)

    print(f"Final Score is {final_score}")

except ValueError as err:
    print(f"Error Occured you give wrong type: {err}")

except:
    print("Try Again and Enter a Proper Value")
finally: 
    print('Finally code is executed ')
```

Nested Error Handling

```python
# use case 

def calculate_sum(x: int, y: int):
    """
    Returns the sum of two numbers.
    """
    return x + y


# Using Nested Error Handling

try:
    print("Welcome to Score Calculator")

    try:
        total_sum = int(input("Enter value for Total Sum: "))
        bonus_points = int(input("Enter value for Bonus Points: "))


        final_score = calculate_sum(total_sum, bonus_points)

        print(f"Final Score is {final_score}")

    except ValueError as err:
        print(f"Oops! Got Error: {err}")
        print("That was not a valid input. Try again...")

except Exception as err:
    print(f"Unexpected Error: {err}")

finally:
    print("Program execution completed.")
```
