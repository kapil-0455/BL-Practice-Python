# Python Operators

Python provides different types of operators:

- **Assignment Operators:** Assign values to variables. (`=`)
- **Arithmetic Operators:** Perform mathematical operations. (`+`, `-`, `*`, `/`, `%`, `**`, `//`)
- **Comparison Operators:** Compare values and return `True` or `False`.
  - `==`, `!=`, `>`, `<`, `>=`, `<=`
- **Logical Operators:** Combine multiple conditions.
  - `and`, `or`, `not`
- **Special Operators:** Check identity or membership.
  - `is`, `is not`, `in`, `not in`

---

# Conditional Operations

- Used to check whether a condition is **True** or **False**.
- Mostly use **comparison**, **logical**, and **special** operators.
- Return a Boolean value (`True` or `False`).

Example:

```python
age >= 18
# Output: True / False
```




````python
# Find Generation code 

birth_year = int(input("Enter Birth Year : "))

if birth_year >= 1946 and birth_year <= 1964:
    print("Generation: Baby Boomer")

elif birth_year >= 1965 and birth_year <= 1980:
    print("Generation: Gen X")

elif birth_year >= 1981 and birth_year <= 1996:
    print("Generation: Millennial")

elif birth_year >= 1997 and birth_year <= 2012:
    print("Generation: Gen Z")

elif birth_year >= 2013 and birth_year <= 2025:
    print("Generation: Gen Alpha")

else:
    print("Generation: Unknown")
````

# Python Shorthand If-Else

- Write an **if-else** statement in a single line.
- Uses the `and` and `or` operators.

**Syntax:**

```python
condition and value_if_true or value_if_false
```

**Example:**

```python
result = gender == "male" and "his" or "her"
```
