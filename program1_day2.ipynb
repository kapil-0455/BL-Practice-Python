## `print()` Function

- The `print()` function is used to display output in Python.
- It writes the given text or value to **standard output (`stdout`)**.
- By default, the output is displayed on the **terminal**, **console**, or **screen**.

### Standard Output (`stdout`)

- **Standard Output (`stdout`)** is the default destination where a program sends its output.
- In most cases, `stdout` refers to the **terminal** or **console**.
- Whenever you use the `print()` function, the output is written to `stdout`.

# Object-Oriented Programming (OOP)

- **Real-World Modeling:** Represents real-world entities using objects.
- **Object:** An instance of a class that represents a real-world entity.
- **Class:** A blueprint for creating objects.
- **Attributes:** Variables that store an object's data.
- **Methods:** Functions that define an object's behavior.
- **Method Overriding:** Redefines a parent class method in a child class.
  - **Example:** Google Colab overrides Python's `print()` to display output in the browser.
