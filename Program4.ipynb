```python
# Good Programming Practice

def calculate_area(length: int, width: int):
    """Return the area of a rectangle."""
    return length * width

length = 10
width = 5

print(f"Area using good practice: {calculate_area(length, width)}")
```

```python
def calculate_total_marks(marks: int, bonus_marks: int) -> int:
    """Return total marks after adding bonus marks."""
    return marks + bonus_marks

student_marks = 90
bonus_marks = 5

print(f"Total marks using good practice: {calculate_total_marks(student_marks, bonus_marks)}")
```
