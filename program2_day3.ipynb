```python
def calculate_monthly_payment(loan, duration, rate):
    monthly_rate = rate / 100 / 12
    months = duration * 12
    payment = (loan * monthly_rate * (1 + monthly_rate) ** months) / ((1 + monthly_rate) ** months - 1)
    return round(payment, 2)

loan = float(input("Enter loan amount: "))
duration = float(input("Enter loan duration (years): "))
rate = float(input("Enter annual interest rate (%): "))

payment = calculate_monthly_payment(loan, duration, rate)

print(f"Monthly payment: {payment}")
```
