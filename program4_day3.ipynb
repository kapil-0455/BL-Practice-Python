```python
#Practice problem 4 find min and max 

def find_max_min(num_lis):
    '''
      Description : finding min and max from list
      Parameter : List 
      Return : mini_ele , max_ele
    '''

    # inbuilt functions
    max_ele = max(num_lis)
    min_ele = min(num_lis)

    # this will return a tuple (min_ele , max_ele)
    return min_ele , max_ele

size = int(input('Enter size of the list : ')) 
num_lis = []

for i in range (size):
    element = int(input('Enter your element : '))
    # adding elements in list
    num_lis.append(element)


# we destructure our tuple
minimum_ele , maximum_ele = find_max_min(num_lis)
print(f'Maximum element in array : {maximum_ele}')
print(f'Minimum element in array : {minimum_ele}')
```
