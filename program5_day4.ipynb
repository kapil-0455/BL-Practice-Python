### Recursion - Print factorial


```python
# Iterative code 
def factorial(num):
    if(num == 0):
        return 1
    else:
        result = 1
        for i in range(1, num + 1):
            result *= i
        return result


number = int(input("Enter the number to find the factorial : "))

result = factorial(number)

print(f"Factorial of {number} is {result}")
```

```python
# Recursive 
def factorial(num):
    if(num == 1):
        return 1
    else:
        return num * factorial(num - 1)


number = int(input("Enter the number to find the factorial : "))
result = factorial(number)
print(f"Factorial of {number} is {result}")

    
```
