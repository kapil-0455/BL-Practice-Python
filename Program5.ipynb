 We made our code interactive taking input from terminal
  - we can use input function to do so
  - input() takes a String as a input 
  - we have to type conversion from string to int
   

```python
# x : int , y : int is only the type hint 
# it dont force python to give only integer
# we can pass anything  like string
def calculate_sum(x: int, y: int):
    """
    Returns the sum of two numbers.
    """
    return x + y

# taking input from user 
# it take input as a string

total_sum = input('Enter value for Total Sum ')
bonus_points = input('Enter value for Bonus Points ')

print(
    f"Data Type of total_sum is {type(total_sum)} \n"
    + f"bonus_points is {type(bonus_points)}"
)

final_score = calculate_sum(total_sum, bonus_points)

print(f"Final Score is {final_score}")
```

We have to convert String to Integer


```python

# Correct code 

def calculate_sum(x: int, y: int):
    """
    Returns the sum of two numbers.
    """
    return x + y

# taking input from user 
# it take input as a string then do type conversion string to Integer

total_sum = int(input('Enter value for Total Sum '))
bonus_points = int(input('Enter value for Bonus Points '))

print(
    f"Data Type of total_sum is {type(total_sum)} \n"
    + f"bonus_points is {type(bonus_points)}"
)

final_score = calculate_sum(total_sum, bonus_points)

print(f"Final Score is {final_score}")
```
