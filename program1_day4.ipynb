## Module
A **Python module** is a **`.py` file** that contains Python code such as **functions, classes, variables, and executable statements**.


## Why Use Modules?
- Organize related code.
- Improve code reusability.
- Make programs easier to maintain.

## Types of Modules

### 1. Custom Modules
- Created by the user.
- Can be imported into other Python programs.

### 2. Built-in Modules
- Predefined modules provided by Python.
- Examples: `os`, `random`, `datetime`, `time`, `sys`, `statistics`, `collections` , `math`.

## random module 

The **`random`** module is a built-in Python module used to generate **random numbers**, **random selections**, and **random arrangements**.

#### How to import ?

```python
#example 
import random
```

## Common functions
### `random.random()`
Returns a random floating-point number between **0.0** and **1.0**.


```python
import random 
random.random()
```

### `random.randint(a, b)`
Returns a random integer between **a** and **b** (both inclusive).


```python
import random 

random.randint(1, 5)
```

### `random.randrange(start, stop, step)`
Returns a random number from the specified range.

```python
import random 
# start from 1 and stop before 5 with jump 2
# give only 1 , 3
random.randrange(1 , 5 , 2)
```

### `random.uniform(a, b)`
Returns a random floating-point number between **a** and **b**.

```python
import random
 
num = random.uniform(1 , 3)
print(f"{num:.2f}")
```

### `random.choice(sequence)`
Returns a random element from a sequence such as a list, tuple, or string.

```python
import random

number_lst = [10 , 20 , 30 , 40]
random_ele = random.choice(number_lst)
print(random_ele)
```

### `random.choices(sequence, k=n)`
Randomly select `one or more items with replacement`. This means the same item can be selected multiple times.

```python
import random

fruits = ["apple", "banana", "mango"]
# k -> number of items to select.
result = random.choices(fruits, k=5)
print(result)
```

### `random.sample(sequence, k)`
Returns **`k unique`** random elements from a sequence **`without replacement`**.

```python
import random

fruits = ["apple", "banana", "mango"]
# k -> number of items to select.
# k should be smaller than length of that list
result = random.sample(fruits, k=2)
print(result)
```

### `random.shuffle(list)`
Randomly shuffles the elements of a list in place.


```python
import random

my_list = [1, 2, 3, 4]

# shuffle in origiinal list
# make changes in original lis 
# return none
new_lst = random.shuffle(my_list)

print(new_lst)
print(my_list)
```

```python
# practice program find probablity of head
import random


def find_probability(num_of_flips):
    if num_of_flips == 0:
        print('Flip atleast once')
        return 
    
    
    head_count = 0
    for _ in range(num_of_flips):
        result = random.choice(['Heads' , 'Tails'])
        if result == 'Heads':
            head_count += 1
    
    percentage_heads = head_count / num_of_flips * 100

    print(f"After flipping the coin {num_of_flips} times, the percentage of times head has come is: {percentage_heads}")
         


num_of_flips = int(input('Enter the number of Flips : '))
find_probability(num_of_flips)
```
